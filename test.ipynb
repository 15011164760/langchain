```python
#导入依赖
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
import os
load_dotenv()

#创建模型（langchain 1.x：参数名是 model，不是 model_name；provider 用 openai）
model = init_chat_model(
    model="qwen-long",
    model_provider="openai",
    api_key=os.getenv("DASHSCOPE_API_KEY"),
    base_url=os.getenv("DASHSCOPE_BASE_URL"),
)

# 测试调用
print(model.invoke("你是谁？").content)
```

```python
response=model.invoke([{"role":"system","content":"请你以医生的口吻回答问题"},{
    "role":"user","content":"我如何保持好的身材呢"
}])
print(response.content)
```

```python
response=model.stream([{"role":"system","content":"请你以老师的口吻回答问题"},{
    "role":"user","content":"中国的人口是多少？"
}])

for message in response:
    print(message.content,end='')
```
