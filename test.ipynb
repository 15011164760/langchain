```python
#导入依赖
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
import os
load_dotenv()

#创建模型（langchain 1.x：参数名是 model，不是 model_name；provider 用 openai）
model = init_chat_model(
    model="qwen-long",
    model_provider="openai",
    api_key=os.getenv("DASHSCOPE_API_KEY"),
    base_url=os.getenv("DASHSCOPE_BASE_URL"),
)

# 测试调用
print(model.invoke("你是谁？").content)
```

```python
response=model.invoke([{"role":"system","content":"请你以医生的口吻回答问题"},{
    "role":"user","content":"今天北京天气怎么样"
}])
print(response.content)
```

```python
response=model.stream([{"role":"system","content":"请你以老师的口吻回答问题"},{
    "role":"user","content":"中国的人口是多少？"
}])

for message in response:
    print(message.content,end='')
```

```python
class Student:
    def __init__(self,name,num):
        self.name=name
        self.num=num
        self.grades={
            "chinese":0,
            "math":0,
            "english":0
        }
    def set_scores(self,type,score):
        if type in self.grades:
           self.grades[type]=score
    def get_scores(self):
        print(f"姓名：{self.name}，学号：{self.num}，的成绩为：")
        for key,value in self.grades.items():
            print(f"{key}：{value}")
student=Student("张三",20210001)
student.set_scores("chinese",95)
student.set_scores("math",95)
student.set_scores("english",95)
student.get_scores()
```

```python
#继承
class Employee:
    def __init__(self,name,num):
        self.name=name
        self.num=num
    def get_info(self):
        print(f"姓名：{self.name}，编号：{self.num}")
class FullTimeEmployee(Employee):
    def __init__(self,name,num,salary):
        super().__init__(name,num)
        self.salary=salary
    def get_info(self):
        print(f"姓名：{self.name}，编号：{self.num}，月薪水：{self.salary}")
class PartTimeEmployee(Employee):
    def __init__(self,name,num,daily_salary,days):
        super().__init__(name,num)
        self.daily_salary=daily_salary
        self.days=days
    def get_info(self):
        print(f"姓名：{self.name}，编号：{self.num}，月薪：{self.daily_salary*self.days}")
FullTimeEmployee1=FullTimeEmployee("张三",20210001,10000)
FullTimeEmployee1.get_info()
PartTimeEmployee1=PartTimeEmployee("李四",20210002,100,20)
PartTimeEmployee1.get_info()
```
